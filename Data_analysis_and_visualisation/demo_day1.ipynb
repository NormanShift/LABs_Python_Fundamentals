## Filter rows

how many listings are entire homes?

mask

```python
listings['room_type'] == 'Entire home/apt'
```

```python
mask = listings['room_type'] == 'Entire home/apt'

print(mask.sum()) # 3760 listings are entire homes/apartments, the sum of a boolean series is the number of True values
print(mask.mean()) # the mean of a boolean series is the proportion of True values
```

mask as a filter

```python
listings[mask].shape
```

Which listings are in Södermalm, and also have room for four or more guests?

```python
listings[listings['neighbourhood_cleansed'] == 'Södermalms' and listings['accommodates'] >= 4]
```

```python
listings[(listings['neighbourhood_cleansed'] == 'Södermalms') & (listings['accommodates'] >= 4)] # & is pandas' version of and, it works on series, not on single values. We need to use parentheses around each condition because & has a higher precedence than ==
```

## Sort, and pick rows and columns

```python
view = ['name', 'neighbourhood_cleansed', 'room_type', 'accommodates', 'number_of_reviews', 'review_scores_rating']

listings[view].sort_values('number_of_reviews', ascending=False).head(5)
```

```python
listings[view].sort_values(['accommodates', 'number_of_reviews'], ascending=[False, False]).head(5) # sort by accommodates descending, then by number_of_reviews ascending
```

```python
listings.nlargest(5, 'accommodates')[view]
```

```python
top = listings[view].sort_values('number_of_reviews', ascending=False).head(5)

print(top.loc[825, 'name']) # loc uses the name
print(top.iloc[0][ 'name']) # iloc uses the position
```
