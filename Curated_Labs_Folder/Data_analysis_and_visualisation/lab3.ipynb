```python
# Task 1 - "Load and look"
# Importing module pandas and using it as alias pd.
# Reading the data (the CSV-file) into the dataframe loans_raw.
# When cleaning a table we never write changes to the original. So we make a duplicate, to then perform the cleanup operations on that copy.
# Using shape on the copy of loans_raw (loans) to get the dimensionality of the dataframe, as rows and columns counts (623 and 9 respectively).

import pandas as pd

loans_raw = pd.read_csv('data/loans.csv')
loans = loans_raw.copy()

loans.shape
```

```python
loans.head(10)
```

```python
loans.info()
# Upon running info on the copy of loans_raw (on loans), we (me actually) can make some observations and draw conclusions from these observations, as follows:
# There are 9 columns and all but one ('notes') are of the string (str or text) datatype (Dtype).
# For the 'notes' column however, it contains no data (NaN or 0 non-null), so I recommend "we" (Me) remove it entirely.
# I would recommend columns 'loan_id', 'member_id', 'loan_date', 'return_date', 'late_fee' be changed to numerical datatypes (int, datetime and float respectively) and 'returned' should be bool (since there is a separeate column for the returned date).
# An additional observation involves 'return_date', which has missing values (30 of them missing out of 623).
```

```python
# Task 2 - "Clean the genre"

# For the column 'genre' there are currently (before any changes) 28 values identified as unique.
loans['genre'].unique()
```

```python
# If not observed at first glance, upon further examination of the categories (or values) under the column 'genre', we see duplicates, the real number of unique categories are 6 when manually observed and counted.
loans['genre'].value_counts()
```

```python
# After examining the column 'genre', we modify the dataframe. Effectively reducint 28 "unique values" down to 9 (still -3 to go).

loans['genre'] = loans['genre'].str.strip().str.lower()

loans['genre'].value_counts()
```

```python
# Using nunique() with print(), we can further observe our modifications (changes) to the copied dataframe. unique() outputs the actual formatting of each "unique" or duplicate value.
print(loans['genre'].nunique())
print(loans['genre'].unique())
```

```python
# Below, two variants producing the same replacements (one of them commented out).

# loans['genre'] = (
#     loans['genre']
#     .replace('science fiction', 'sci-fi')
#     .replace('sci fi', 'sci-fi')
#     .str.replace("Children's", "Children")
#     .str.title()
# )

loans['genre'] = loans['genre'].replace({'sci fi': 'sci-fi', 'science fiction': 'sci-fi', "children's": 'children'}).str.title()
 

print(loans['genre'].unique())
```
