**Task 2 - Clean the genre**

```python
# For the column 'genre' there are currently (before any changes) 28 values identified as unique.
loans['genre'].unique()
```

```python
# If not observed at first glance, upon further examination of the categories (or values) under the column 'genre', we see duplicates, the real number of unique categories are 6 when manually observed and counted.
loans['genre'].value_counts()
```

```python
# After examining the column 'genre', we modify the dataframe. Effectively reducint 28 "unique values" down to 9 (still -3 to go).

loans['genre'] = loans['genre'].str.strip().str.lower()

loans['genre'].value_counts()
```

```python
# Using nunique() with print(), we can further observe our modifications (changes) to the copied dataframe. unique() outputs the actual formatting of each "unique" or duplicate value.
print(loans['genre'].nunique())
print(loans['genre'].unique())
```

```python
# Below, two variants producing the same replacements (one of them commented out).

# loans['genre'] = (
#     loans['genre']
#     .replace('science fiction', 'sci-fi')
#     .replace('sci fi', 'sci-fi')
#     .str.replace("Children's", "Children")
#     .str.title()
# )

loans['genre'] = loans['genre'].replace({'sci fi': 'sci-fi', 'science fiction': 'sci-fi', "children's": 'children'}).str.title()
 
# In the printed output we can see that we are down to unique 6 categories (genres).

print(loans['genre'].unique())
```

```python
# Loans for each category or genre listed:
# crime       163
# children    135
# fantasy      90
# sci-fi       90
# romance      86
# history      59

loans['genre'].value_counts()
```

**Task 3 - Yes and no**

```python
# Examining unique values for the 'returned' column. 10 unique values identified where there should be only 2 (bool).
loans['returned'].unique()
```

```python
# Remapping string values to bool values for the column 'returned'.
loans['returned'] = loans['returned'].str.lower().map({'Yes':True, 'YES':True, 'Y':True, 'y':True, 'yes':True, 'No':False, 'N':False, 'n':False, 'NO':False, 'no':False})
```

```python
# Confirming that our remapping was successful.
loans['returned'].unique()
```

```python
# Further examination of our remapping.
print(loans['returned'])
```

**Task 5 - Dates**

```python
l_date = pd.to_datetime(loans['loan_date'], errors='coerce') # Invalid parsing will be set as NaT.
print(loans['loan_date'].isna().sum())
print(l_date.isna().sum())

loans.loc[l_date.isna(), ['loan_id','loan_date']]
```

```python
loans['loan_date'] = l_date

print(loans['loan_date'].dtype)
print(loans['loan_date'].min())
print(loans['loan_date'].max())
print(loans['loan_date'].max() - loans['loan_date'].min())
```
