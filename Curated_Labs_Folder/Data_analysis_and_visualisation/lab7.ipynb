Part 1 - Task 1:8: Set up, and the first histogram

```python
# Importing the relevand libraries.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Applying Seaborn's darkgrid theme.
sns.set_theme(style='darkgrid')

# Creating a python list containing column names with dates for parsing.
date_columns = ['last_scraped', 'calendar_last_scraped', 'first_review', 'last_review']

# Loading the CSV and parsing the date_columns for the proper datatype in the DataFrame.
listings = pd.read_csv('data/listings_clean.csv', parse_dates=date_columns)

# Displaying the number of rows and columns in the DataFrame.
print(listings.shape)
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(listings['number_of_reviews'].dropna(), bins=50)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

sns.histplot(data=listings, x='number_of_reviews', bins=50, ax=ax)

ax.set_title('Distribution of Reviews per Listing: Many Low Counts and a Long Right Tail')
ax.set_xlabel('Number of reviews')
ax.set_ylabel('Number of listings')

plt.show()
```

```python
# Confirming that a high number of listings has 0 reviews.
print(listings['number_of_reviews'].value_counts().head())
```

Part 1 - Task 2:8: Mean, median and the fence

```python
# The column to be analyzed.
reviews = listings['number_of_reviews']

# Calculating the median and mean for 'number_of_reviews'.
reviews_median = reviews.median()
reviews_mean = reviews.mean()

# Calculating the first and third quartiles.
q1 = reviews.quantile(0.25)
q3 = reviews.quantile(0.75)

# Calculating the interquartile range and upper fence.
iqr = q3 - q1
upper_fence = q3 + 1.5 * iqr

print(f'Median: {reviews_median:.1f}')
print(f'Mean: {reviews_mean:.1f}')
print(f'Upper fence: {upper_fence:.1f}')
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

sns.histplot(data=listings, x='number_of_reviews', bins=50, ax=ax)

# Adding vertical reference lines.
ax.axvline(reviews_median, color='black', linestyle='--', label=f'median {median:,.0f}')
ax.axvline(reviews_mean, color='red', linestyle='--', label=f'mean {mean:,.0f}')
ax.axvline(upper_fence, color='gray', linestyle='--', label=f'upper fence {upper:,.0f}')

ax.set_title('High Review Counts Pull the Mean right of the Median')
ax.set_xlabel('Number of reviews')
ax.set_ylabel('Number of listings')
ax.legend()

plt.show()
```

```python
print((listings['number_of_reviews'] > upper_fence).value_counts())
print("Object with max number of reviews:", listings['number_of_reviews'].max())
```

**The upper fence is furthest to the right at 69.5 reviews. The mean, approximately 35.8, is higher than the median of 9 because the distribution is strongly right-skewed. A relatively small number of listings with very high review counts pull the mean upward, while the median is less affected by these extreme values.**
